```python
class TextDataset(Dataset):
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __len__(self): return len(self.x)
    def __getitem__(self, idx):
        return self.x[idx], self.y[idx]
 
dataset = TextDataset(inputs, targets)
loader = DataLoader(dataset, batch_size=8, shuffle=True)
```

```python
class MiniTransformerEncoder(nn.Module):
    def __init__(self, vocab_size, d_model=64, d_ff=128, seq_len=6):
        super().__init__()
        self.seq_len = seq_len
        self.embed = nn.Embedding(vocab_size, d_model)
        self.pos_enc = self.sinusoidal_encoding(seq_len, d_model)
 
        self.ln1 = nn.LayerNorm(d_model)
        self.attn = nn.MultiheadAttention(d_model, num_heads=1, batch_first=True)
        self.ln2 = nn.LayerNorm(d_model)
        self.ffn = nn.Sequential(
            nn.Linear(d_model, d_ff),
            nn.ReLU(),
            nn.Linear(d_ff, d_model)
        )
        self.final_ln = nn.LayerNorm(d_model)
        self.out = nn.Linear(d_model, vocab_size)
 
    def sinusoidal_encoding(self, seq_len, d_model):
        pos = torch.arange(seq_len).unsqueeze(1)
        i = torch.arange(d_model).unsqueeze(0)
        angle_rates = 1 / torch.pow(10000, (2 * (i // 2)) / d_model)
        angle_rads = pos * angle_rates
        pe = torch.zeros(seq_len, d_model)
        pe[:, 0::2] = torch.sin(angle_rads[:, 0::2])
        pe[:, 1::2] = torch.cos(angle_rads[:, 1::2])
        return pe.unsqueeze(0)
 
    def forward(self, x, return_attention=False):
        # x shape: (batch, seq_len)
        x = self.embed(x) + self.pos_enc[:, :x.size(1), :].to(x.device)
    
        # Attention block
        attn_out, attn_weights = self.attn(self.ln1(x), self.ln1(x), self.ln1(x))
        x = x + attn_out
    
        # Feed-forward block
        ffn_out = self.ffn(self.ln2(x))
        x = x + ffn_out
    
        # Final layer norm + projection
        x = self.final_ln(x)
        logits = self.out(x)
    
        if return_attention:
            return logits, attn_weights
        return logits
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print(device)
```

```python
model = MiniTransformerEncoder(vocab_size=vocab_size).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=3e-4)
 
epochs = 25
 
for epoch in range(epochs):
    total_loss = 0
    model.train()
    for xb, yb in loader:
        xb, yb = xb.to(device), yb.to(device)
 
        optimizer.zero_grad()
        logits = model(xb)  # (batch, seq_len, vocab_size)
 
        loss = criterion(logits.view(-1, vocab_size), yb.view(-1))
        loss.backward()
        optimizer.step()
 
        total_loss += loss.item()
 
    if (epoch + 1) % 5 == 0:
        print(f"Epoch {epoch+1}/{epochs} - Loss: {total_loss/len(loader):.4f}")
```

```python
model.eval()
question = ["the", "cat", "reads"]
x = torch.tensor([encode(question)], dtype=torch.long).to(device)

for _ in range(400):
    with torch.no_grad():

        x_input = x[:, -model.seq_len:]  # truncate to last `seq_len` tokens
        logits = model(x_input)

        temperature = 0.8
        probs = torch.softmax(logits[0, -1] / temperature, dim=-1)
        next_token = torch.multinomial(probs, num_samples=1)

        x = torch.cat([x, next_token.view(1, 1)], dim=1)


print("\n🌤️ Generated sequence:")
print(decode(x[0].tolist()))
```
